## Outline
Here, we explore pootting a Gaussian 'peak'. Showing the first and second derivatives of this.  
Then we add noise to teh Gaussian and show the affect thathas on these derivatives. 

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.signal import savgol_filter
from scipy.stats import norm
```

## Generate a Gaussian

```python
x = np.arange(-5, 5, 0.1) # x values from -5 to 5
y = norm.pdf(x, 0, 1)       

plt.plot(x, y)
plt.title('Gaussian (Normal) Distribution (mean=0, std=1)')
plt.show()
```

## First derivative

```python
dx = x[1] - x[0]
Zf = savgol_filter(y, window_length=9, polyorder=4, deriv=1, delta=dx)
plt.plot(x, Zf)
plt.title('First derivative')
plt.show()
```

## Second derivative

```python
dx = x[1] - x[0]
Zf = savgol_filter(y, window_length=9, polyorder=4, deriv=2, delta=dx)
plt.plot(x, Zf)
plt.title('Second derivative')
plt.show()
```

## Add noise

```python
# Define noise parameters
mean = 0
std_dev = 0.01  # Adjust this to control noise intensity (sigma)

# 3. Generate Gaussian noise with the same shape as the profile
noise = np.random.normal(mean, std_dev, size=y.shape)

# 4. Add noise to the profile
y_noisy = y + noise

# Optional: Clip values if your profile must remain within specific bounds (e.g., [0, 1])
# y_noisy = np.clip(y_noisy, 0, 1)

plt.plot(x, y_noisy)
plt.title('Noisy Gaussian (Normal) Distribution (mean=0, std=1)')
plt.show()
```

## First derivative of noisy Gaussian

```python
dx = x[1] - x[0]
Zf = savgol_filter(y_noisy, window_length=9, polyorder=4, deriv=1, delta=dx)
plt.plot(x, Zf)
plt.title('First derivative of noisy Gaussian')
plt.show()
```

## Second derivative of noisy Gaussian

```python
dx = x[1] - x[0]
Zf = savgol_filter(y_noisy, window_length=9, polyorder=4, deriv=2, delta=dx)
plt.plot(x, Zf)
plt.title('Second derivative of noisy Gaussian')
plt.show()
```
